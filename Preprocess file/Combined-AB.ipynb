```python
import os
import shutil
import re

def extract_number(filename):
    """Extracts the digits from the filename for proper sorting."""
    match = re.search(r'\d+', filename)
    return int(match.group()) if match else 0

def merge_sequences(src_dir1, src_dir2, dest_dir):
    # Create destination directory if it doesn't exist
    os.makedirs(dest_dir, exist_ok=True)
    
    # Get and sort files from both directories
    files_data1 = sorted([f for f in os.listdir(src_dir1) if f.startswith('Sub') and f.endswith('.npy')], key=extract_number)
    files_data2 = sorted([f for f in os.listdir(src_dir2) if f.startswith('Sub') and f.endswith('.npy')], key=extract_number)
    
    current_index = 1
    
    # Process data1
    print(f"Processing files from {src_dir1}...")
    for file_name in files_data1:
        src_path = os.path.join(src_dir1, file_name)
        new_name = f"Sub_{current_index:03d}.npy"
        dest_path = os.path.join(dest_dir, new_name)
        
        shutil.copy2(src_path, dest_path)
        current_index += 1

    # Process data2
    print(f"Processing files from {src_dir2}...")
    for file_name in files_data2:
        src_path = os.path.join(src_dir2, file_name)
        new_name = f"Sub_{current_index:03d}.npy"
        dest_path = os.path.join(dest_dir, new_name)
        
        shutil.copy2(src_path, dest_path)
        current_index += 1

    print(f"Done! All files merged into '{dest_dir}'. Total files: {current_index - 1}")

# --- Configuration ---
DATA1_FOLDER = "/teamspace/studios/this_studio/phdResearch/data1n/Subjects/"
DATA2_FOLDER = "/teamspace/studios/this_studio/phdResearch/data2n/Subjects/"
OUTPUT_FOLDER = "/teamspace/studios/this_studio/phdResearch/data12n/Subjects/"

merge_sequences(DATA1_FOLDER, DATA2_FOLDER, OUTPUT_FOLDER)
```

```python
import os
import pandas as pd

# Read CSV files
df1 = pd.read_csv("/teamspace/studios/this_studio/phdResearch/data1n/Labels/labels.csv")
df2 = pd.read_csv("/teamspace/studios/this_studio/phdResearch/data2n/Labels/labels.csv")

max_id = (
    df1["subject_id"]
    .str.extract(r'(\d+)')
    .astype(int)
    .max()
    .iloc[0]
)

# Update subject IDs in the second file
nums = df2["subject_id"].str.extract(r'(\d+)').astype(int)[0]
df2["subject_id"] = nums.add(max_id).map(lambda x: f"Sub_{x:03d}")

# Merge
merged = pd.concat([df1, df2], ignore_index=True)

# Save
os.makedirs("/teamspace/studios/this_studio/phdResearch/data12n/Labels", exist_ok=True)
merged.to_csv("/teamspace/studios/this_studio/phdResearch/data12n/Labels/labels.csv", index=False)

print(merged.head())
print(merged.tail())
```
